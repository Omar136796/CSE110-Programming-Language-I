```python
#task 4
def function_name(str1):
    countL=0
    countU=0
    for i in str1:
        if 97<=ord(i)<=122:
            countL+=1
        if 65<=ord(i)<=90:
            countU+=1
    print(f"No. of Uppercase characters : {countU}")
    print(f"No. of Lowercase Characters: {countL}")
function_name("The quick Sand Man")
```

```python
#task 5
def calculate_tax(age,salary,designation):
    tax=0
    if age<18:
        tax=0
    elif ("president" in designation) or ("President" in designation):
        tax=0
    elif salary<10000:
        tax=0
    elif 10000<salary<20000:
        tax=salary*(5/100)
    else:
        tax=salary*(10/100)
    return tax
age=int(input())
salary=int(input())
designation=input()
print(calculate_tax(age,salary,designation))
```

```python
#task 7
def show_palindrome(num,separate=""):
    s=""
    for i in range(1,num+1):
        s+=str(i)+separate
    for j in range(num-1,0,-1):
        s+=str(j)+separate
    return s
print(show_palindrome(5))
```

```python
#task 8
def show_palindromic_triangle(num):
    for i in range(1,num+1):
        print(" "*(2*(num-i)),end="")
        print(show_palindrome(i," "),end="")
        print()
show_palindromic_triangle(6)
```

```python
#task 9
def area_circumference_generator(num):
    pi=3.141592653589793
    area=pi*(num**2)
    circumference=2*pi*num
    list1=[]
    list1.append(area)
    list1.append(circumference)
    tup1=tuple(list1)
    return tup1
tup1=area_circumference_generator(1.5)
a,c=tup1
print(tup1)
print(f"Area of the circle is {a} and circumference is {c}")
```

```python
#task 10
def make_square(tup1):
    num1=tup1[0]
    num2=tup1[-1]
    dict1={}
    val=0
    for i in range(num1,num2+1):
        val=i**2
        dict1[i]=val
    return dict1
print(make_square((1,3)))
```

```python
#task 11
def rem_duplicate(tup1):
    list1=list(tup1)
    list2=[]
    for i in list1:
        if i not in list2:
            list2.append(i)
    tup2=tuple(list2)
    return tup2
print(rem_duplicate(("Hi", 1, 2, 3, 3, "Hi",'a', 'a', [1,2])))
```

```python
#task 13
def function_name(operator,num1,num2):
    output=0
    if operator=="+":
        output=num1+num2
    if operator=="-":
        output=num1-num2
    if operator=="*":
        output=num1*num2
    if operator=="/":
        output=num1/num2
    return output
operator=input()
operator=operator[1:-1:]
num1=float(input())
num2=float(input())
print(function_name(operator,num1,num2))
```

```python
#task 14
def function_name(sentence,position):     #"I love programming.", 3
    str1=""
    str2=""
    str3=""
    for i in range(len(sentence)):
        if i==0:
            str1+=sentence[i]
        elif i%position!=0:
            str1+=sentence[i]
        else:
            str2+=sentence[i]
    str3=str1+str2
    return str3
sentence=input()
sentence=sentence[1:-1:]
position=int(input())
print(function_name(sentence,position))
```
