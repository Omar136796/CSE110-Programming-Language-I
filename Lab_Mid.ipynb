```python
#coding problem
inp1=input()     #ksGSvsRE
str1=""          #PSp
countu=0
countl=0
for i in inp1:
    if 65<=ord(i)<=90:
        countu+=1
    else:
        countl+=1
for j in inp1:
    if countl>=countu:
        if 97<=ord(j)<=122:
            str1+=chr(ord(j)-4)
        else:
            str1+=chr(ord(j)+28)
    elif countl<countu:
        if 97<=ord(j)<=122:
            str1+=chr(ord(j)-36)
        else:
            str1+=chr(ord(j)-4)
print(str1)
```
