```python
#task 8
inp1=input()    #Jon:100, Dan:200, Rob:300
list1=[i for i in inp1.split(", ")]
dict1={}
for i in list1:
    list2=[j for j in i.split(":")]    # i[1:-1:] dile (on:10) ashe
    key=list2[0]
    val=int(list2[-1])
    dict1[key]=val
total=0
count=0
for v in dict1.values():
    total+=v
    count+=1
print(f"Average is {total//count}.")
```

```python
#task 11
inp1=input()       #"Python programming is fun"
inp1=inp1[1:-1:].lower()
dict1={}
for i in inp1:
    if i == " ":
        continue
    else:
        if i not in dict1:
            dict1[i]=1
        else:
            dict1[i]+=1
print(dict1)
```

```python
inp1=input()   #2 5 8 4
list1=[int(i) for i in inp1.split(" ")]
print("Input list:",list1)
even_list=[]
for i in list1:
    if (int(i))%2==0:
        even_list.append(i)
print("Even list:",even_list)
print("Length of even list:",len(even_list))
dict1={}
if len(even_list)==0:
    print("No dictionary can be made")
else:
    for j in range(len(even_list)):
        x=input()
        dict1[x]=even_list[j]
print(dict1)
max=0
key=""
for k,v in dict1.items():
    if int(v)>max:
        max=v
        key=k
print(key)
```
