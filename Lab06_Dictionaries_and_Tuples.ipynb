```python
#task 8
inp1=input() #Jon:100, Dan:200, Rob:300
list1=inp1.split(",")
dict1={}
for i in list1:
    list2=i.split(":")   # list2.split()  na i.split()
    key=list2[0]
    val=int(list2[-1])
    dict1[key]=val
total=0
count=0
for v in dict1.values():
    count+=1
    total+=int(v)
print("Average is",total//count)
```

```python
#task 9
exam_marks={'Cierra Vega': 175, 'Alden Cantrell': 200, 'Kierra Gentry': 165, 'Pierre Cox': 190}
num=int(input())
dict1={}
for k,v in exam_marks.items():
    if v>=num:
        dict1[k]=v
    else:
        continue
print(dict1)
```

```python
#task 10
dict1={'sci fi': 5, 'mystery': 3, 'horror': 14, 'young_adult': 2, 'adventure':9}
max=0
key=""
for k,v in dict1.items():
    if int(v)>max:
        max=int(v)
        key=k
print(f"The highest selling book genre is '{key}' and the number of books sold are {max}.")
```

```python
#task 11
inp1=input()       #"Python programming is fun"
inp1=inp1[1:-1:].lower()
dict1={}
for i in inp1:
    if i == " ":
        continue
    else:
        if i not in dict1:
            dict1[i]=1
        else:
            dict1[i]+=1
print(dict1)
```
